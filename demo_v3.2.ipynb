```python
# test the HuggingFaceEmbeddings
## 测试环境有没有安装好(如果)

from langchain.embeddings import HuggingFaceEmbeddings #从 langchain.embeddings类 导入 HuggingFaceEmbeddings 的函数
import numpy as np

embeddings = HuggingFaceEmbeddings() #利用默认的模型做编码嵌入
text = "This is a test document." #给定一个字符串
text2 = 'You are so good.' #给定另一个字符串

query_result = embeddings.embed_query(text) #str进行编码
query_result2 = embeddings.embed_query(text2) #str进行编码
doc_result = embeddings.embed_documents([text, text2]) #list进行编码

print('The query_result result:',query_result)
print('The doc_result result:',doc_result)
```

```python
from langchain.document_loaders import PyPDFLoader ## 加载读取PDF的函数-》PyPDFLoader

# Create an instance of PyPDFLoader Class, input is the path of the pdf document to be loaded
loader = PyPDFLoader("./Regulation.pdf") #输入文件路径

# call PyPDFLoader Class function load to load pdf files
#docs = []
docs = loader.load() ##读取PDF文件

print(f'documents:{len(docs)}')
```

```python
print('The one page length:',len(docs[9].page_content))
print('Print one page text:',docs[9].page_content)
```

```python
# Split docs
from langchain.text_splitter import RecursiveCharacterTextSplitter, CharacterTextSplitter 
##加载了两种文件切分方式，分别是 RecursiveCharacterTextSplitter，CharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 200,  # The size of each text block.
    chunk_overlap = 20,  # The overlap between each text block.
    #separator=["\n\n", "\n", " ", "","."]
)

text_splitter2 = CharacterTextSplitter(
    chunk_size = 200,  # The size of each text block.
    chunk_overlap = 20,  # The overlap between each text block.
    separator="." #, "\n", " ","","."]
)

split_docs = text_splitter.split_documents(docs)
print("split_docs size:",len(split_docs))

split2_docs = text_splitter2.split_documents(docs)
print("split2_docs size:",len(split2_docs))
```

```python
## 在已经建立好向量化的知识库后，只需要首先读取加载这库

from langchain.vectorstores import Chroma ## 调用 chromadb (向量化数据库)
from langchain.embeddings.huggingface import HuggingFaceEmbeddings
import sentence_transformers
from langchain.embeddings import HuggingFaceEmbeddings #从 langchain.embeddings类 导入 HuggingFaceEmbeddings 的函数
import numpy as np

embedding_model_dict = {
    "ernie-tiny": "nghuyong/ernie-3.0-nano-zh",
    "ernie-base": "nghuyong/ernie-3.0-base-zh",
    "text2vec": "GanymedeNil/text2vec-large-chinese",
    "text2vec2":"uer/sbert-base-chinese-nli",
    "text2vec3":"shibing624/text2vec-base-chinese",
}

EMBEDDING_MODEL = "text2vec3"

# 初始化 hugginFace 的 embeddings 对象
embeddings = HuggingFaceEmbeddings() #model_name=embedding_model_dict[EMBEDDING_MODEL]
embeddings.client = sentence_transformers.SentenceTransformer(embeddings.model_name, device='cpu')

vectordb_docs = Chroma(persist_directory='./docs/chroma/', embedding_function=embeddings)
vectordb_docs
```

```python
# 提出问题
question_ = "what is 'Unit and Study Load'？" 

# 根据相似度给出检索答案
docs_ans = vectordb_docs.similarity_search(question_,k=3) # k是代表返回相似性最高的Top-K查询结果

# 打印检索结果
for i in range(len(docs_ans)):
    print('The answer of docs:\n',docs_ans[i])
```
